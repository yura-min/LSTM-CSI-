```python
import scipy.io as sio
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
import matplotlib.pyplot as plt
from google.colab import drive
import os

# 1) 데이터 로드 (.mat)
# 1) Drive 마운트
drive.mount('/content/drive')
mat_path = "/content/drive/MyDrive/csi_dataset.mat"

data = sio.loadmat(mat_path)

csi = data["csi"]        # (N, 64, 1, 1) complex
rx_pos = data["rx_pos"]  # (N, 3)

N, F, _, _ = csi.shape   # N: 샘플 수, F=64 서브캐리어

# 2) csi -> real / imag + 위치 붙이기
csi = csi.reshape(N, F)           # (N, 64) complex
csi_real = np.real(csi)           # (N, 64)
csi_imag = np.imag(csi)           # (N, 64)

# 한 시점 feature: [real64, imag64, x, y, z] = 131차원
features = np.concatenate(
    [csi_real, csi_imag, rx_pos], axis=1
)   # (N, 131)

# 3) train / test 시간 순 분할 (90 / 10)
split_percentage = 0.9
split_point = round(N * split_percentage)

train_feat = features[:split_point]
test_feat = features[split_point:]

# 4) 스케일링 (train으로만 fit)
scaler = MinMaxScaler()
scaler.fit(train_feat)

scaled_train = scaler.transform(train_feat)
scaled_test  = scaler.transform(test_feat)

# 5) 시계열 윈도우 만들기 (lags = L)
def timeseries_preprocessing(scaled_train, scaled_test, lags):
    # train
    X_tr, Y_tr = [], []
    for t in range(len(scaled_train) - lags - 1):
        # 입력: t ~ t+lags-1 까지 L개 시점의 전체 feature(131)
        X_tr.append(scaled_train[t : t+lags, :])
        # 출력: t+lags 시점의 CSI(real+imag)만 (앞 128차원)
        Y_tr.append(scaled_train[t+lags, :128])

    # test
    X_te, Y_te = [], []
    for t in range(len(scaled_test) - lags - 1):
        X_te.append(scaled_test[t : t+lags, :])
        Y_te.append(scaled_test[t+lags, :128])

    X_train = np.array(X_tr)   # (samples, lags, 131)
    Y_train = np.array(Y_tr)   # (samples, 128)
    X_test  = np.array(X_te)
    Y_test  = np.array(Y_te)

    return X_train, Y_train, X_test, Y_test

lags = 10  # 주가 예제처럼 10 스텝 사용
X_train, Y_train, X_test, Y_test = timeseries_preprocessing(
    scaled_train, scaled_test, lags
)

print("X_train shape:", X_train.shape)  # (?, 10, 131)
print("Y_train shape:", Y_train.shape)  # (?, 128)

# 6) LSTM 모델 정의
n_features = X_train.shape[2]  # 131

model = Sequential()
model.add(LSTM(256, input_shape=(lags, n_features)))
model.add(Dense(128))  # 다음 시점 CSI(real+imag)

model.compile(optimizer='adam', loss='mse')

```

```python
# 7) 학습
history = model.fit(
    X_train, Y_train,
    epochs=50,
    validation_data=(X_test, Y_test),
    shuffle=False
)

# 8) 학습/검증 손실 그래프
plt.plot(history.history['loss'], label='loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.legend()
plt.title("CSI LSTM training")
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn import metrics
import numpy as np
import matplotlib.pyplot as plt

# 9) scaler에서 CSI 부분(앞 128차원)만 떼서 y용 scaler 만들기
scaler_csi = MinMaxScaler()
scaler_csi.min_        = scaler.min_[:128].copy()
scaler_csi.scale_      = scaler.scale_[:128].copy()
scaler_csi.data_min_   = scaler.data_min_[:128].copy()
scaler_csi.data_max_   = scaler.data_max_[:128].copy()
scaler_csi.data_range_ = scaler.data_range_[:128].copy()
scaler_csi.n_features_in_ = 128
scaler_csi.feature_range = scaler.feature_range  # (0,1) 그대로

# 10) 예측 (스케일된 공간에서)
Y_pred_scaled = model.predict(X_test)   # (n_samples, 128)

# 11) 실제 단위 CSI로 되돌리기
Y_pred = scaler_csi.inverse_transform(Y_pred_scaled)  # (n_samples, 128)
Y_true = scaler_csi.inverse_transform(Y_test)         # (n_samples, 128)

# ----------------------------
# A. 서브캐리어 하나의 "실수부" 플롯 (Medium 스타일)
# ----------------------------
subc = 0   # 0번 서브캐리어, 필요하면 다른 index로 바꿔보기

true_real = Y_true[:, subc]      # 0~63: real
pred_real = Y_pred[:, subc]

plt.figure(figsize=(8,4))
plt.plot(true_real, label=f'True real (sc {subc})')
plt.plot(pred_real, label=f'Pred real (sc {subc})')
plt.legend()
plt.title(f'CSI real-part prediction (subcarrier {subc})')
plt.xlabel('Test sample index')
plt.ylabel('Real(h)')
plt.show()

# ----------------------------
# B. 같은 서브캐리어의 "진폭 |h|" 플롯
# ----------------------------
true_real_all = Y_true[:, :64]
true_imag_all = Y_true[:, 64:]
pred_real_all = Y_pred[:, :64]
pred_imag_all = Y_pred[:, 64:]

mag_true = np.sqrt(true_real_all[:, subc]**2 + true_imag_all[:, subc]**2)
mag_pred = np.sqrt(pred_real_all[:, subc]**2 + pred_imag_all[:, subc]**2)

plt.figure(figsize=(8,4))
plt.plot(mag_true, label=f'True |h| (sc {subc})')
plt.plot(mag_pred, label=f'Pred |h| (sc {subc})')
plt.legend()
plt.title(f'CSI magnitude prediction (subcarrier {subc})')
plt.xlabel('Test sample index')
plt.ylabel('|h|')
plt.show()

# ----------------------------
# C. 전체 CSI 벡터 기준으로 평가 지표
# ----------------------------
mae = metrics.mean_absolute_error(Y_true, Y_pred)
mse = metrics.mean_squared_error(Y_true, Y_pred)
rmse = np.sqrt(mse)
r2 = metrics.r2_score(Y_true, Y_pred)

print("Prediction performance on CSI (real+imag 128-dim)")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R2  :", r2)
```
